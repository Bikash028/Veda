Step 1 — Load the Dataset

```python
import pandas as pd

df = pd.read_csv("Superstore.csv")

print(df.head())
print(df.shape)
```

```python
print(df.info())
```

Step 2: Missing Value Audit

```python
# Check missing values
missing_values = df.isnull().sum()

print(missing_values)
```

Step 3: Duplicate Record Audit

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
duplicate_percentage = (
    duplicate_count / len(df) * 100
)

print(f"Duplicate percentage: {duplicate_percentage:.2f}%")
```

```python
duplicate_report = pd.DataFrame({
    "Issue": ["Duplicate Rows"],
    "Count": [duplicate_count],
    "Percentage": [round(duplicate_percentage, 2)]
})

print(duplicate_report)
```

Step 4: Range Validation

```python
# check saales
invalid_sales = (df["Sales"] <= 0).sum()

print("Invalid Sales:", invalid_sales)
```

```python
# check quantity
invalid_quantity = (df["Quantity"] <= 0).sum()

print("Invalid Quantity:", invalid_quantity)
```

```python
invalid_discount = (
    (df["Discount"] < 0) | (df["Discount"] > 1)
).sum()

print("Invalid Discount:", invalid_discount)
```

```python
print("Invalid Sales:", invalid_sales)
print("Invalid Quantity:", invalid_quantity)
print("Invalid Discount:", invalid_discount)
```

Step 5: Data Type & Format Consistency Audit

```python
print(df.dtypes)
```

```python
print("Order Date type:", df["Order Date"].dtype)
print("Ship Date type:", df["Ship Date"].dtype)
```

```python
print(df[["Order Date", "Ship Date"]].dtypes)
```

Step 6: Consistency Audit

```python
text_columns = [
    "Ship Mode",
    "Segment",
    "Country",
    "City",
    "State",
    "Region",
    "Category",
    "Sub-Category"
]

for col in text_columns:
    spaces = df[col].astype(str).str.strip().ne(
        df[col].astype(str)
    ).sum()
    
    print(f"{col}: {spaces} values with extra spaces")
```

```python
for col in ["Ship Mode", "Segment", "Region", "Category"]:
    print(f"\n--- {col} ---")
    print(df[col].unique())
```

```python
invalid_dates = (
    df["Ship Date"] < df["Order Date"]
).sum()

print("Ship Date before Order Date:", invalid_dates)
```

Step 7: Create the Cleaned Dataset

```python
cleaned_df = df.copy()
```

```python
cleaned_df = cleaned_df.drop_duplicates()

print("Original rows:", len(df))
print("Cleaned rows:", len(cleaned_df))
print("Rows removed:", len(df) - len(cleaned_df))
```

Clean text columns

```python
text_columns = [
    "Ship Mode",
    "Segment",
    "Country",
    "City",
    "State",
    "Region",
    "Category",
    "Sub-Category"
]

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].astype(str).str.strip()
```

Ensure dates are properly formatted

```python
cleaned_df["Order Date"] = pd.to_datetime(
    cleaned_df["Order Date"],
    errors="coerce"
)

cleaned_df["Ship Date"] = pd.to_datetime(
    cleaned_df["Ship Date"],
    errors="coerce"
)
```

Check the cleaned dataset

```python
print(cleaned_df.info())
```

```python
print(cleaned_df.head())
```

Final quality check

```python
print("Remaining duplicates:",
      cleaned_df.duplicated().sum())

print("\nMissing values:")
print(cleaned_df.isnull().sum())
```

```python
cleaned_df.to_csv(
    "Day_24_Cleaned_Superstore.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```

```python
# Final Data Quality Audit Summary

total_rows = len(df)
total_columns = len(df.columns)

total_missing = df.isnull().sum().sum()
duplicate_count = df.duplicated().sum()

invalid_sales = (df["Sales"] <= 0).sum()
invalid_quantity = (df["Quantity"] <= 0).sum()

invalid_discount = (
    (df["Discount"] < 0) |
    (df["Discount"] > 1)
).sum()

invalid_order_dates = df["Order Date"].isna().sum()
invalid_ship_dates = df["Ship Date"].isna().sum()

invalid_date_order = (
    df["Ship Date"] < df["Order Date"]
).sum()

total_issues = (
    total_missing +
    duplicate_count +
    invalid_sales +
    invalid_quantity +
    invalid_discount +
    invalid_order_dates +
    invalid_ship_dates +
    invalid_date_order
)

audit_summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Columns",
        "Total Missing Values",
        "Duplicate Rows",
        "Invalid Sales",
        "Invalid Quantity",
        "Invalid Discount",
        "Invalid Order Dates",
        "Invalid Ship Dates",
        "Ship Date Before Order Date",
        "Total Detected Issues"
    ],
    "Count": [
        total_rows,
        total_columns,
        total_missing,
        duplicate_count,
        invalid_sales,
        invalid_quantity,
        invalid_discount,
        invalid_order_dates,
        invalid_ship_dates,
        invalid_date_order,
        total_issues
    ]
})

print(audit_summary)
```
